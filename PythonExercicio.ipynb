
**Um dispositivo para ensaios de tração com capacidade máxima de 220 000 N. Tem duas barras de sustentação cilíndricas, cada uma
suporta metade da carga máxima. Devem ser usadas barras de aço-carbono(1045) lixadas e polidas; a resistência ao escoamento e o
limite de resistência à tração mínimos são de 310MPa e 565MPa, respectivamente**


Tensão de trabalho é tensão de escoamento / fator de segurança

Tensão de trabalho é força / área

fator de segurança [N = 5]


```python
import math

print("Esse programa irá cslcular o custo de uma barra")

print("Essa etapa irá calcular a tensão de trabalho (MPa)")

Tensao_de_escoamento = float(input("Digite a tensão de escoamento em (MPa) "))

Fator_de_segurança = float(input("Digite o fator de segurança "))

Tensao_de_trabalho = Tensao_de_escoamento / Fator_de_segurança

print("A tensão de trabalho é ",Tensao_de_trabalho," em MPa")
```


Área = força / tensão de trabalho
* pi r^2 = força / tensão de trabalho

* pi (d/2)^2 = força / tensão de trabalho

* diâmetro = raiz(força / tensão de trabalho pi)2

* diâmetro = raiz(110 000 N / 62 10^6 N/m^2 pi)2

* diâmetro = 0,047528 m

```python
print("Essa etapa irá calcular o diâmetro da barra em função de uma força")

Força = float(input("Digite a força em (N) "))

Numero_da_barra = float(input("Digite quantas barra "))

Força_por_n_de_barra = Força / Numero_da_barra

Tensao_de_trabalho_em_N_m2 = Tensao_de_trabalho * 10 ** 6

Diametro_m = (math.sqrt(Força_por_n_de_barra / (Tensao_de_trabalho_em_N_m2 * math.pi))) * 2

print("O diâmetro da barra é ",Diametro_m," em metros")

Diametro_cm = ((math.sqrt(Força_por_n_de_barra / (Tensao_de_trabalho_em_N_m2 * math.pi))) * 2) * 100

print("O diâmetro da barra é ",Diametro_cm," em centímetros")

Diametro_mm = ((math.sqrt(Força_por_n_de_barra / (Tensao_de_trabalho_em_N_m2 * math.pi))) * 2) * 1000

print("O diâmetro da barra é ",Diametro_mm," em milímetros")
```

Considerando comprimento de 30 cm e densidade de 7,8 g/com^3,determine a massa em kg

Volume cilíndro = Area da base x altura

Volume = pi (diâmetro/2)^2 x altura

Volume = pi (4,7528 cm/2)^2 x 30 cm

Volume = 532,6913 cm^3



```python
print("Essa etapa irá calcular o volume da barra")

Altura_da_barra = float(input("Digite a altura da barra em (cm) "))

Volume_da_barra = ((math.pi * (Diametro_cm / 2) ** 2) * Altura_da_barra)

print("O volume da barra é ",Volume_da_barra," (cm³) ")

print("Essa etapa irá calcular a massa da barra")

Densidade = float(input("Digite a densidade da barra em (g/cm³) "))

print("A densidade é ", Densidade," (g/cm³) ")

Massa_da_barra = Volume_da_barra * Densidade

Massa_da_barra_kg = Massa_da_barra / 1000

print("A massa da barra é ", Massa_da_barra," g")

print("A massa da barra é ", Massa_da_barra_kg," kg")

custo_por_quilo = float(input("Digite o custo por quilo em R$/Kg "))

Custo_total_da_barra = round(Massa_da_barra_kg * custo_por_quilo, 2)

print("O custo da barra é ", Custo_total_da_barra, "R$")

Custo_total = Custo_total_da_barra * Numero_da_barra

print("O custo total é ", Custo_total, "R$")
```
