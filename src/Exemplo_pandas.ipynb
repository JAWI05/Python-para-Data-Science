```python
# Dada a hora e os minutos qual será o horário depois de um determinado intervalo de tempo
hora = int(input('Digite a hora: '))
minutos = int(input('Digite os minutos: '))
duracao = int(input('Digite a o intervalo de tempo em minutos: '))
if duracao < 60:
  minutos += duracao
  if minutos >= 60:
    hora += minutos // 60
    minutos = minutos % 60
if duracao > 60:
  hora += duracao // 60
  minutos += duracao % 60
  if minutos >= 60:
    hora += minutos // 60
    minutos = minutos % 60
if hora >= 24:
  hora = hora % 24
print(hora, minutos, sep=':')
```
