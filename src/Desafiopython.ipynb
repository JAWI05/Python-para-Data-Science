```python
import math
import numpy as np
import matplotlib.pyplot as plt

# Dados dos materiais: módulo de elasticidade (GPa), densidade (g/cm³), custo por quilo (R$/kg)
materiais = {
    "Aço 1045": {"Elasticidade": 205, "densidade": 7.85, "custo": 7},
    "Epóxi e fibra de aramida": {"Elasticidade": 70, "densidade": 1.44, "custo": 80},
    "Epóxi e fibra de carbono": {"Elasticidade": 230, "densidade": 1.55, "custo": 160},
    "Epóxi e fibra de vidro": {"Elasticidade": 72, "densidade": 2.5, "custo": 25},
    "Aço inoxidável 304": {"Elasticidade": 193, "densidade": 8.0, "custo": 15},
    "Aço 1020": {"Elasticidade": 205, "densidade": 7.85, "custo": 6},
    "Epóxi": {"Elasticidade": 3.5, "densidade": 1.2, "custo": 35}
}

# Fatores de segurança
fatores_de_seguranca = [2.0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.0]

# Dados da força e altura da barra
forca = 220_000  # N
altura_da_barra_cm = 30  # cm
altura_da_barra_m = altura_da_barra_cm / 100  # Convertendo para metros

# Função para calcular o diâmetro, volume, massa e custo
def calcular_custos(material, densidade, custo_por_quilo, tensao_trabalho):
    tensao_trabalho_pa = tensao_trabalho * 1e6  # Convertendo para Pa
    diametro_m = (math.sqrt(forca / (tensao_trabalho_pa * math.pi))) * 2  # Diâmetro em metros
    diametro_cm = diametro_m * 100  # Convertendo para centímetros
    volume_cm3 = math.pi * (diametro_cm / 2) ** 2 * altura_da_barra_cm  # Volume em cm³
    massa_g = volume_cm3 * densidade  # Massa em gramas
    massa_kg = massa_g / 1000  # Convertendo para kg
    custo_total = massa_kg * custo_por_quilo  # Custo total
    return custo_total

# Matriz para armazenar os custos
custos_matriz = np.zeros((len(materiais), len(fatores_de_seguranca)))

# Preenchendo a matriz de custos para cada material e fator de segurança
for i, (material, propriedades) in enumerate(materiais.items()):
    for j, fator in enumerate(fatores_de_seguranca):
        tensao_escoamento = propriedades["Elasticidade"]  # Utilizando o módulo de elasticidade como tensão de escoamento para simplificação
        tensao_trabalho = tensao_escoamento / fator
        custo_total = calcular_custos(material, propriedades["densidade"], propriedades["custo"], tensao_trabalho)
        custos_matriz[i, j] = custo_total

# Criando o gráfico de mapa de calor
fig, ax = plt.subplots()
im = ax.imshow(custos_matriz, cmap="coolwarm")

# Nomear os eixos com os materiais e fatores de segurança
ax.set_xticks(np.arange(len(fatores_de_seguranca)))
ax.set_yticks(np.arange(len(materiais)))

ax.set_xticklabels(fatores_de_seguranca)
ax.set_yticklabels(materiais)

# Rotacionar os rótulos no eixo x
plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")

# Inserir os valores no mapa de calor
for i in range(len(materiais)):
    for j in range(len(fatores_de_seguranca)):
        text = ax.text(j, i, round(custos_matriz[i, j], 2), ha="center", va="center", color="black")

# Título e ajustes finais
ax.set_title("Custo das barras (R$) em função do fator de segurança")
fig.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Definir os dados dos materiais
materiais = {
    'Material': [
        'Aço 1045',
        'Epóxi e Fibra de Aramida',
        'Epóxi e Fibra de Carbono',
        'Epóxi e Fibra de Vidro',
        'Aço Inoxidável 304',
        'Aço 1020'
    ],
    'E_GPa': [210, 70, 230, 72, 193, 205],          # Módulo de Elasticidade em GPa
    'Densidade_kg_m3': [7850, 1600, 1600, 2500, 8000, 7850],  # Densidade em kg/m³
    'Custo_por_kg_USD': [0.8, 15, 20, 10, 2.0, 0.6],          # Custo por kg em USD
    'Sigma_yield_MPa': [530, 1000, 1500, 900, 215, 350]      # Tensão de Escoamento em MPa
}

df_materiais = pd.DataFrame(materiais)

# 2. Definir os fatores de segurança
fatores_seguranca = [2, 2.5, 3, 3.5, 4, 4.5, 5.0]

# 3. Definir a força e a altura da barra
F = 220000  # Força em N
L = 0.3     # Altura da barra em metros (30 cm)

# 4. Converter unidades
df_materiais['E_Pa'] = df_materiais['E_GPa'] * 1e9            # GPa para Pa
df_materiais['Sigma_yield_Pa'] = df_materiais['Sigma_yield_MPa'] * 1e6  # MPa para Pa

# 5. Criar DataFrames para armazenar os resultados
resultados_area = pd.DataFrame(index=df_materiais['Material'], columns=fatores_seguranca)
resultados_peso = pd.DataFrame(index=df_materiais['Material'], columns=fatores_seguranca)
resultados_custo = pd.DataFrame(index=df_materiais['Material'], columns=fatores_seguranca)

# 6. Calcular Área, Peso e Custo para cada combinação de Material e Fator de Segurança
for _, material in df_materiais.iterrows():
    for sf in fatores_seguranca:
        # Calcular a área necessária (A = F * SF / sigma_yield)
        A = F * sf / material['Sigma_yield_Pa']  # m²

        # Calcular o peso (W = densidade * A * L)
        W = material['Densidade_kg_m3'] * A * L  # kg

        # Calcular o custo (C = W * custo_por_kg)
        C = W * material['Custo_por_kg_USD']    # USD

        # Armazenar os resultados
        resultados_area.loc[material['Material'], sf] = A
        resultados_peso.loc[material['Material'], sf] = W
        resultados_custo.loc[material['Material'], sf] = C

# Converter os resultados para tipo float
resultados_area = resultados_area.astype(float)
resultados_peso = resultados_peso.astype(float)
resultados_custo = resultados_custo.astype(float)

# 7. Exibir os DataFrames (opcional)
print("Área Necessária (m²):")
print(resultados_area)
print("\nPeso (kg):")
print(resultados_peso)
print("\nCusto (USD):")
print(resultados_custo)

# 8. Plotar os mapas de calor
plt.figure(figsize=(21, 6))

# Mapa de Calor para Área Necessária
plt.subplot(1, 3, 1)
sns.heatmap(resultados_area, annot=True, fmt=".2e", cmap='viridis')
plt.title('Área Necessária (m²)')
plt.ylabel('Material')
plt.xlabel('Fator de Segurança')

# Mapa de Calor para Peso
plt.subplot(1, 3, 2)
sns.heatmap(resultados_peso, annot=True, fmt=".2f", cmap='magma')
plt.title('Peso (kg)')
plt.ylabel('Material')
plt.xlabel('Fator de Segurança')

# Mapa de Calor para Custo
plt.subplot(1, 3, 3)
sns.heatmap(resultados_custo, annot=True, fmt=".2f", cmap='coolwarm')
plt.title('Custo (USD)')
plt.ylabel('Material')
plt.xlabel('Fator de Segurança')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import seaborn as sns

# melhor visualização
# módulo de elasticidade (gPa), densidade (kg/m^3) e custo por quilograma(kg) (R$/kg)
materiais = {
    'Aço 1045': {'Elasticidade': 210*1e9, 'densidade': 7850, 'custo': 5.50},
    'Epóxi e Aramida': {'Elasticidade': 70*1e9, 'densidade': 1400, 'custo': 50.00},
    'Epóxi e Carbono': {'Elasticidade': 230*1e9, 'densidade': 1600, 'custo': 150.00},
    'Epóxi e Vidro': {'Elasticidade': 85*1e9, 'densidade': 2000, 'custo': 60.00},
    'Aço Inox 304': {'Elasticidade': 200*1e9, 'densidade': 8000, 'custo': 20.00},
    'Aço 1020': {'Elasticidade': 205*1e9, 'densidade': 7870, 'custo': 4.50},
    'Epóxi': {'Elasticidade': 3.5*1e9, 'densidade': 1200, 'custo': 30.00}
}

# Dados constante:
forca = 220000 # N
altura = 0.30 # m
fatores_de_seguranca = [2, 2.5, 3, 3.5, 4, 4.5, 5.0]

def calcular_custo(material, fator_seguranca):
    Elasticidade = materiais[material]['Elasticidade']  # Pa
    densidade = materiais[material]['densidade']  # kg/m^3
    custo_por_kg = materiais[material]['custo']  # R$/kg

    # valor máximo de tensão que um material resiste sem danos
    tensao_admissivel = Elasticidade / fator_seguranca

    area = forca / tensao_admissivel

    volume = area * altura

    massa = volume * densidade

    return massa

    custo_total = massa * custo_por_kg

    return custo_total

# Gerar a matriz de custos
custos = np.zeros((len(materiais), len(fatores_de_seguranca)))

for i, material in enumerate(materiais):
    for j, fator_seguranca in enumerate(fatores_de_seguranca):
        custos[i, j] = calcular_custo(material, fator_seguranca)

# Visualizar o mapa de calor
sns.heatmap(custos, annot=True, xticklabels=fatores_de_seguranca, yticklabels=materiais.keys(), cmap="rocket_r")
plt.title("Custo de barras por material e fator de segurança (R$)")
plt.xlabel("Fator de Segurança")
plt.ylabel("Material")
plt.show()
```
