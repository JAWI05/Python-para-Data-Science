```python
from pydrive2.auth import GoogleAuth
from pydrive2.drive import GoogleDrive
from google.colab import auth
from oauth2client.client import GoogleCredentials
```

```python
# Autentifique e crie o cliente do PyDrive2
auth.authenticate_user()
gauth = GoogleAuth()
gauth.credentials = GoogleCredentials.get_application_default()
drive = GoogleDrive(gauth)
```

```python
# Crie e faça upload de um arquivo de texto
uploaded = drive.CreateFile({'title': 'Python Data Science.txt'})
uploaded.SetContentString('Python aplicado a Data Science')
uploaded.Upload()
print('Uploaded file with ID {}'.format(uploaded.get('id')))
```

```python
# Use o id do arquivo para carregá-lo e imprime seu conteúdo
downloaded = drive.CreateFile({'id': uploaded.get('id')})
print('Dowloaded content "{}"'.format(downloaded.GetContentString))
```

```python
# Para usar a API REST do Drive, primeiro é necessário autentificar e constrir um cliente da API

from google.colab import auth
auth.authenticate_user()
from googleapiclient.discovery import build
drive_service = build('drive', 'v3')

# Com esse cliente é possível usar todas as funções da Referência da API do Google Drive

with open('/tmp/to_upload.txt', 'w') as f:
  f.write('Python aplicado a Data Science, Percy Jackson')

print('/tmp/to_upload.txt contains:')
!cat /tmp/to_upload.txt

# Faça upload usando o método files.create
from googleapiclient.http import MediaFileUpload
file_metadata = {
    'name': 'You are a wizard, harry.txt',
    'mimetype' : 'text/plain'
}

media = MediaFileUpload('/tmp/to_upload.txt',
                        mimetype='text/plain',
                        resumable=True)
file = drive_service.files().create(body=file_metadata,
                                    media_body=media,
                                    fields='id').execute()
print('File ID: {}'.format(file.get('id')))
```
