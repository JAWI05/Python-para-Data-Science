```python
print('Esse programa irá ler a média de 5 estudantes e calcula a média da turma')

contador = 1
acumulador_notas = 0

while contador <= 5:
  media = float(input('Digite a nota do estudante: '))
  contador += 1
  acumulador_notas += media

media_turma = acumulador_notas / (contador - 1)
print('A média da turma é ', media_turma)
```

```python
print('Esse programa lê 10 números inteiros e diz a quantidade de pares e ímpares')
contador = 1
contador_pares = 0
contador_impares = 0

while contador <= 10:
  num = int(input('Digite um número inteiro: '))
  if num % 2 == 0:
    contador_pares += 1
  else:
    contador_impares += 1
  contador += 1

print('A quantidade de pares é ', contador_pares)
print('A quantidade de ímpares é ', contador_impares)
```
