```python
print('Esse algoritmo lê um número, se for maior que 20, divide por 2 e mostra o resultado')
num = int(input('Digite um número: '))
if num > 20:
  num = num/2
  print('A metade do número é ', num)
```

```python
print('Esse programa lê a idade e diz se a pessoa é maior de idade ou não')
idade = int(input('Digite a sua idade: '))
if idade >= 18:
  print('Você é maior de idade')
else:
  print('Você é menor de idade')
```

```python
print('Esse programa lê dois números. Mostra se forem iguais, se forem diferentes imprimi a diferença do maior para o menor')
num1 = int(input('Digite um número: '))
num2 = int(input('Digite outro número: '))
if num1 == num2:
  print('Os números são iguais')
else:
  if num1 > num2:
    dif = num1 - num2
    print('A diferença entre os dois números é ',  dif)
  else:
    dif = num2 - num1
    print('A diferença entre os dois números é ', dif)
```

```python
print('Esse programa lê dois números. Imprimi em ordem crescente')
numf1 = float(input('Digite um número: '))
numf2 = float(input('Digite outro número, diferente do primeiro: '))
if numf1 < numf2:
  print('Em ordem: ', numf1, ' - ', numf2)
else:
  print('Em ordem: ', numf2, ' - ', numf1)
```

```python
print('Esse programa lê o salário e a quantidade de anos de cada funcionário e calcula o valor de bônus')
salario = float(input('Digite o seu salário '))
anos = int(input('Digite quantos anos você trabalha na empresa '))
if(anos >= 5):
  bonus = salario * 0.2
  print('Você receberá ', bonus, 'de bônus')
  salario = salario + bonus
  print('Seu novo salário é ', salario)
else:
  bonus = salario * 0.1
  print('Você receberá ', bonus, 'de bônus')
  salario = salario + bonus
  print('Seu novo salário é ', salario)
```

```python
print('O algoritmo recebe o salário e o valor da prestação e informa se o empréstimo pode ou não ser concedido')
salariobruto = float(input('Digite o seu salário bruto '))
Valor_prestacao = float(input('Digite o valor da prestação '))
porc30 = salariobruto * 0.3
if Valor_prestacao >= porc30:
  print('Como ', Valor_prestacao, '>=', porc30)
  print('Empréstimo não pode ser concedido')
else:
  print('Como ', Valor_prestacao, '<', porc30)
  print('Empréstimo concedido')
```
